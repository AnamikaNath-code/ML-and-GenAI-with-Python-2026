```python
#function to print first 10 natural numbers
def natural_num():
    for i in range(1,11):
        print(i, end=" ")

natural_num()
```

```python
#function for sum of natural numbers
def sum_natural(n):
    s=0
    for i in range(1,n+1):
      s=s+i
    print("Sum of",n,"natural numbers: ",s)
sum_natural(11)
```

```python
#function to reverse a number
def rev_num(num):
  s=0
  while num>0:
    a=num%10
    s=s*10+a
    num=num//10
  print("Reversed number: ",s)
num=int(input("Enter a number: "))
rev_num(num)
```

```python
#function to count digits in a number
def count_digit(num):
  count=0
  while num>0:
    num=num//10
    count=count+1
  print("Number of digits: ",count)

count_digit(98765432)
count_digit(9)
```

```python
#function to check palindrome number
def palindrome(num):
  original_num=num
  rev_num=0
  while num>0:
    a=num%10
    rev_num=rev_num*10+a
    num=num//10
  if original_num==rev_num:
    print(original_num ,"is palindrome")
  else:
    print(original_num ,"is not palindrome")
palindrome(121)
palindrome(61991)
```

```python
#function for fibonacci series
def fibonacci(n):
  a,b=0,1
  if n==1:
    print(a)
  elif n==2:
    print(a,b,end=" ")
  else:
    print(a,b,end=" ")
    for i in range(2,n):
      z=a+b
      a,b=b,z
      print(z,end=" ")

  print('\n')

fibonacci(7)
fibonacci(2)
```

```python
#calculator
def calculator():
    print("********** CALCULATOR **********")
    print("\n")

    print("MENU")
    print("1.Addition")
    print("2.Subtraction")
    print("3.Multiplication")
    print("4.Division")
    print("5.Exit")

    while True:
      ch=int(input("Enter your choice: "))
      if ch==1:
        a=int(input("Enter first number: "))
        b=int(input("Enter second number: "))
        print("Addition: ",a+b)
      elif ch==2:
        a=int(input("Enter first number: "))
        b=int(input("Enter second number: "))
        print("Subtraction: ",a-b)
      elif ch==3:
        a=int(input("Enter first number: "))
        b=int(input("Enter second number: "))
        print("Multiplication: ",a*b)
      elif ch==4:
        a=int(input("Enter first number: "))
        b=int(input("Enter second number: "))
        print("Division of first by second number: ",a/b)
      elif ch==5:
        print("Exit")
        break
      else:
        print("Invalid choice")

calculator()
```

```python
#handling zero division error
def zerodiv():
    try:
      a=int(input("Enter numerator: "))
      b=int(input("Enter denominator: "))
      c=a/b
      print("Division: ",c)
    except ZeroDivisionError:
      print("Error: Division by zero not possible")

zerodiv()
zerodiv()
```
